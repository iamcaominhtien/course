## Chuẩn bị

````python
import numpy as np
import pandas as pd
import torch

print("PyTorch", torch.__version__)

# Tensor dùng xuyên suốt lab
A = torch.tensor([[11, 12, 13],
                  [21, 22, 23],
                  [31, 32, 33]])
````

## Bài 1. Tạo tensor 2D và đọc thuộc tính

Tạo `B` có **2 hàng, 4 cột** chứa các số 1 đến 8 theo thứ tự, điền từng hàng một:
```
[[1, 2, 3, 4],
 [5, 6, 7, 8]]
```
Lưu `shape_B` (kích thước), `rank_B` (số chiều) và `count_B` (tổng số phần tử).

````python
B = torch.tensor([[1, 2, 3, 4], [5, 6, 7, 8]])
shape_B = B.shape
rank_B = B.ndimension()
count_B = B.numel()
print(B, shape_B, rank_B, count_B)
````

````python
assert B.tolist() == [[1, 2, 3, 4], [5, 6, 7, 8]]
assert tuple(shape_B) == (2, 4) and rank_B == 2 and count_B == 8
print("✅ Bài 1 đúng")
````

## Bài 2. Index và slice trên `A`

Với `A` ở ô chuẩn bị, lấy:
- `elem`: phần tử ở **hàng thứ 2, cột thứ 3** (nhớ index từ 0);
- `row2`: cả hàng thứ 2;
- `col3`: cả cột thứ 3;
- `sub`: ma trận con gồm **hai hàng cuối** và **hai cột đầu**.

````python
elem = A[1, 2]
row2 = A[1, :]
col3 = A[:, 2]
sub = A[1:3, 0:2]
print(elem, row2, col3, sub, sep="\n")
````

````python
assert elem.item() == 23
assert row2.tolist() == [21, 22, 23]
assert col3.tolist() == [13, 23, 33]
assert sub.tolist() == [[21, 22], [31, 32]]
print("✅ Bài 2 đúng")
````

## Bài 3. Bẫy: `A[1:3][...]` khác `A[1:3, ...]`

Mục tiêu: lấy **cột thứ 3 của hai hàng cuối**, tức `[23, 33]`.
- (a) Viết đúng vào `right`.
- (b) Thử `A[1:3][2]` trong `try/except`; lưu **tên lớp lỗi** (ví dụ `"IndexError"`) vào `err_name`.
- (c) `looks_ok = A[1:3][1]`. Đoán trước: nó là hàng nào của `A`? Chạy để kiểm tra.

````python
right = A[1:3, 2]

err_name = None
try:
    A[1:3][2]
except IndexError as e:
    err_name = type(e).__name__

looks_ok = A[1:3][1]
print(right, err_name, looks_ok)
````

````python
assert right.tolist() == [23, 33]
assert err_name == "IndexError"
assert looks_ok.tolist() == [31, 32, 33], "A[1:3][1] là hàng thứ 3 của A, KHÔNG phải cột"
print("✅ Bài 3 đúng")
````

## Bài 4. Sửa cả một khối bằng slice

Tạo `A2 = A.clone()`. Đặt **cột thứ 2 của hai hàng cuối** về 0 bằng **một lệnh gán**.
Kiểm tra: `A` gốc không đổi.

````python
A2 = A.clone()
A2[1:3, 1] = 0
print(A2)
````

````python
assert A2.tolist() == [[11, 12, 13], [21, 0, 23], [31, 0, 33]]
assert A.tolist() == [[11, 12, 13], [21, 22, 23], [31, 32, 33]], "A gốc không được đổi"
print("✅ Bài 4 đúng")
````

## Bài 5. Cộng, nhân vô hướng, Hadamard

`P = [[1, 0], [0, 1]]`, `Q = [[2, 1], [1, 2]]`. Tính:
- `added`: `P + Q`;
- `doubled`: `Q` nhân 2;
- `hadamard`: nhân từng phần tử của `P` với `Q`.

````python
P = torch.tensor([[1, 0], [0, 1]])
Q = torch.tensor([[2, 1], [1, 2]])
added = P + Q
doubled = 2 * Q
hadamard = P * Q
print(added, doubled, hadamard, sep="\n")
````

````python
assert added.tolist() == [[3, 1], [1, 3]]
assert doubled.tolist() == [[4, 2], [2, 4]]
assert hadamard.tolist() == [[2, 0], [0, 2]]
print("✅ Bài 5 đúng")
````

## Bài 6. Nhân ma trận và quy tắc kích thước

- (a) `M` shape `(2, 3)` = `[[0, 1, 1], [1, 0, 1]]`, `N` shape `(3, 2)` = `[[1, 1], [1, 1], [-1, 1]]`. Tính `prod = M` nhân ma trận `N`. **Dự đoán shape trước.**
- (b) Tạo `a = torch.ones(4, 3)`, `b = torch.ones(3, 5)`. Lưu `out_shape = tuple((a @ b).shape)`.
- (c) Thử `b @ a` trong `try/except RuntimeError`, lưu thông báo vào `msg`. Vì sao lỗi?

````python
M = torch.tensor([[0, 1, 1], [1, 0, 1]])
N = torch.tensor([[1, 1], [1, 1], [-1, 1]])
prod = M @ N

a = torch.ones(4, 3)
b = torch.ones(3, 5)
out_shape = tuple((a @ b).shape)

msg = None
try:
    b @ a
except RuntimeError as e:
    msg = str(e)
print(prod, out_shape, msg, sep="\n")
````

````python
assert prod.tolist() == [[0, 2], [0, 2]]
assert out_shape == (4, 5)
assert msg is not None, "b @ a phải báo lỗi: (3,5) @ (4,3) có 5 != 4"
print("✅ Bài 6 đúng")
````

## Bài 7. Từ bảng Pandas đến tensor, rồi chuẩn hóa từng cột

`df` là bảng 4 người, 2 đặc trưng (chiều cao cm, cân nặng kg). Hàng = mẫu, cột = đặc trưng.
- `X`: tensor `float32` lấy từ `df`, shape `(4, 2)`.
- `Xs`: chuẩn hóa **từng cột**: `(X - trung bình cột) / độ lệch chuẩn cột`.
  Sau chuẩn hóa mỗi cột có trung bình ≈ 0 và độ lệch chuẩn ≈ 1.

````python
df = pd.DataFrame({"height": [150, 160, 170, 180], "weight": [50, 60, 80, 90]})
X = torch.tensor(df.values, dtype=torch.float32)
Xs = (X - X.mean(dim=0)) / X.std(dim=0)
print(X.shape, Xs.mean(dim=0), Xs.std(dim=0))
````

````python
assert X.shape == (4, 2) and X.dtype == torch.float32
assert torch.allclose(Xs.mean(dim=0), torch.zeros(2), atol=1e-5)
assert torch.allclose(Xs.std(dim=0), torch.ones(2), atol=1e-5)
print("✅ Bài 7 đúng")
````

## Bài 8. Tensor 3D: ảnh màu nhỏ

`img = torch.zeros(2, 2, 3)` là ảnh 2×2 pixel, 3 kênh (R, G, B), theo dạng (cao, rộng, kênh).
- Tô pixel ở hàng 0, cột 1 thành **đỏ thuần** `[1, 0, 0]` bằng một lệnh gán.
- `chw`: đổi thứ tự trục sang kiểu PyTorch `(kênh, cao, rộng)`.
- `red`: kênh đỏ của `chw` (tensor 2×2).

````python
img = torch.zeros(2, 2, 3)
img[0, 1] = torch.tensor([1., 0., 0.])
chw = img.permute(2, 0, 1)
red = chw[0]
print(chw.shape)
print(red)
````

````python
assert tuple(chw.shape) == (3, 2, 2)
assert red.tolist() == [[0.0, 1.0], [0.0, 0.0]]
assert chw[1].sum().item() == 0 and chw[2].sum().item() == 0, "kênh G, B phải bằng 0"
print("✅ Bài 8 đúng")
````

## Bài 9. Thử thách: một lớp tuyến tính viết tay

Một lớp fully connected tính `out = X @ W + b`.
- `X`: 4 mẫu, mỗi mẫu 3 đặc trưng, toàn số 1: shape `(4, 3)`.
- `W = [[1, 0], [0, 1], [1, 1]]`: shape `(3, 2)` (kiểu float).
- `b = [0.5, -0.5]`: shape `(2,)`.

Tính `out` bằng **một biểu thức**, rồi trả lời: shape của `out` là gì và `b` được cộng vào thế nào khi kích thước khác nhau?

````python
X = torch.ones(4, 3)
W = torch.tensor([[1., 0.], [0., 1.], [1., 1.]])
b = torch.tensor([0.5, -0.5])
out = X @ W + b
print(out.shape)
print(out)
````

````python
assert tuple(out.shape) == (4, 2)
assert out.tolist() == [[2.5, 1.5]] * 4
print("✅ Bài 9 đúng: đây chính là phép tính của một lớp Linear")
````
