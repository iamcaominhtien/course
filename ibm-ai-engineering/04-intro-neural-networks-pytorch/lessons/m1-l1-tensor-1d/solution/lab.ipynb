## Bài 1. Kiểu dữ liệu (dtype)

Tạo `ints` từ list `[1, 2, 3]` và `floats` từ list `[1.0, 2.0, 3.0]`.
**Trước khi chạy**, hãy đoán `dtype` của từng tensor. Sau đó lưu `ints.dtype` vào `ints_dtype` và `floats.dtype` vào `floats_dtype`.

```python
ints = torch.tensor([1, 2, 3])
floats = torch.tensor([1.0, 2.0, 3.0])
ints_dtype = ints.dtype
floats_dtype = floats.dtype
print(ints_dtype, floats_dtype)
```

## Bài 2. Đổi kiểu và cái bẫy của `mean()`

Cho `scores = torch.tensor([2, 4, 6, 8])`.
1. Tạo một ô riêng, gọi `scores.mean()` và đọc thông báo lỗi.
2. Tạo `scores_f`: bản **số thực float32** của `scores` (không được sửa `scores` gốc).
3. Tính `avg`: trung bình của `scores_f`.

## Bài 3. Hình dạng và `view`

`x = torch.arange(6)` là tensor `[0, 1, 2, 3, 4, 5]`.
Từ `x`, dùng `view` tạo:
- `col`: dạng **cột**, shape `(6, 1)`, và dùng `-1` ít nhất một lần;
- `mat`: shape `(2, 3)`.

Câu hỏi suy nghĩ: `view` có sao chép dữ liệu không? (Ô kiểm tra sẽ cho bạn biết.)

```python
x = torch.arange(6)
print(x.shape, x.numel())
col = x.view(-1, 1)
mat = x.view(2, 3)
print(col.shape, mat.shape)
```

```python
assert col.shape == (6, 1)
assert mat.shape == (2, 3)
assert col.data_ptr() == x.data_ptr(), "view không sao chép: dùng chung bộ nhớ với x"
print("✅ Bài 3 đúng (và view dùng chung bộ nhớ với tensor gốc)")
```

## Bài 4. Index, slice và việc dùng chung bộ nhớ

Cho `t = torch.tensor([10, 20, 30, 40, 50, 60])`.
- (a) `last3`: ba phần tử cuối của `t`, lấy bằng **slice**.
- (b) `t2 = t.clone()`, rồi đặt vị trí 1 và 4 của `t2` thành 0 **bằng một lệnh gán** dùng list index.
- (c) **Dự đoán trước:** sau `last3[0] = -1`, tensor `t` có bị đổi không? Lưu kết quả thật (`True`/`False`) vào `changed`.

```python
t = torch.tensor([10, 20, 30, 40, 50, 60])
t_original = t.clone()

last3 = t[-3:]

t2 = t.clone()
t2[[1, 4]] = 0

last3[0] = -1
changed = not torch.equal(t, t_original)
print(t, t2, changed)
```

```python
assert t2.tolist() == [10, 0, 30, 40, 0, 60], "t2 chưa đúng"
assert changed is True, "slice là view nên t phải bị đổi"
assert t.tolist() == [10, 20, 30, -1, 50, 60]
print("✅ Bài 4 đúng: slice là view, sửa slice là sửa tensor gốc")
```

## Bài 5. NumPy: dùng chung hay sao chép?

`arr = np.array([1.0, 2.0, 3.0])`.
- `shared`: tensor tạo bằng `torch.from_numpy`.
- `copied`: tensor **độc lập** với `arr`.

Sau khi hai tensor được tạo, đoạn code đặt `arr[:] = 0`. Yêu cầu: `shared` phải về toàn số 0, còn `copied` vẫn là `[1, 2, 3]`.

```python
arr = np.array([1.0, 2.0, 3.0])
shared = torch.from_numpy(arr)
copied = torch.tensor(arr)
arr[:] = 0
print(shared, copied)
```

```python
assert shared.tolist() == [0.0, 0.0, 0.0], "shared phải đổi theo arr"
assert copied.tolist() == [1.0, 2.0, 3.0], "copied phải độc lập với arr"
print("✅ Bài 5 đúng")
```

## Bài 6. Pandas Series → tensor

## Bài 7. Hàm thống kê

`w = torch.tensor([3.0, -1.0, 4.0, 1.0, 5.0])`. Tính:
- `mean_w`: trung bình;
- `std_w`: độ lệch chuẩn;
- `max_w`: giá trị lớn nhất;
- `argmax_idx`: **vị trí** (index) của giá trị lớn nhất.

```python
w = torch.tensor([3.0, -1.0, 4.0, 1.0, 5.0])
mean_w = w.mean()
std_w = w.std()
max_w = w.max()
argmax_idx = w.argmax()
print(mean_w, std_w, max_w, argmax_idx)
```

```python
assert abs(float(mean_w) - 2.4) < 1e-5
assert abs(float(std_w) - 2.4083) < 1e-3
assert float(max_w) == 5.0
assert int(argmax_idx) == 4
print("✅ Bài 7 đúng")
```

## Bài 8. Bốn phép toán nền tảng

`u = [1., 2., 3.]`, `v = [4., 5., 6.]`. Tính:
- `added`: `u + v`;
- `scaled`: `u` nhân với 3;
- `hadamard`: nhân từng phần tử `u` với `v`;
- `dotp`: tích vô hướng của `u` và `v`.

**Dự đoán trước** `hadamard` có shape gì, `dotp` có shape gì?

```python
u = torch.tensor([1., 2., 3.])
v = torch.tensor([4., 5., 6.])
added = u + v
scaled = 3 * u
hadamard = u * v
dotp = torch.dot(u, v)
print(added, scaled, hadamard, dotp)
```

```python
assert added.tolist() == [5., 7., 9.]
assert scaled.tolist() == [3., 6., 9.]
assert hadamard.tolist() == [4., 10., 18.]
assert dotp.item() == 32.0 and dotp.dim() == 0, "dot phải ra một số (0 chiều)"
print("✅ Bài 8 đúng")
```

## Bài 9. Broadcasting và lỗi kích thước

Dùng `u = [1., 2., 3.]`.
- (a) `shifted`: cộng 100 vào mọi phần tử của `u` (một dòng, không vòng lặp).

## Bài 11. Thử thách: độ tương đồng cosine

Viết hàm `cosine(u, v)` trả về cosine của góc giữa hai vector:
`dot(u, v) / (độ dài u × độ dài v)`, với độ dài (norm) của vector là căn bậc hai của tổng bình phương các phần tử.

Nó dùng tích vô hướng để đo hai vector "cùng hướng" đến mức nào (xem lại mục 2 của `lesson.md`).
Kết quả mong đợi: cùng hướng → 1, vuông góc → 0, ngược hướng → -1.

```python
def cosine(u, v):
    return torch.dot(u, v) / (u.norm() * v.norm())
```

```python
a = torch.tensor([1., 2., 3.])
assert abs(cosine(a, 2 * a).item() - 1.0) < 1e-5, "cùng hướng → 1"
assert abs(cosine(torch.tensor([1., 0.]), torch.tensor([0., 1.])).item()) < 1e-6, "vuông góc → 0"
assert abs(cosine(a, -a).item() + 1.0) < 1e-5, "ngược hướng → -1"
print("✅ Bài 11 đúng")
```

## Bài 12. Thử thách: chuẩn hóa min-max

Viết hàm `minmax(x)` đưa mọi giá trị về đoạn [0, 1]: `(x - min) / (max - min)`.
Gặp tensor số nguyên thì hàm phải vẫn chạy được (hãy nghĩ cách xử lý kiểu dữ liệu).

```python
def minmax(x):
    x = x.float()
    return (x - x.min()) / (x.max() - x.min())
```

```python
out = minmax(torch.tensor([2., 4., 6.]))
assert out.tolist() == [0.0, 0.5, 1.0]
out_int = minmax(torch.tensor([10, 20, 30]))
assert out_int.tolist() == [0.0, 0.5, 1.0], "tensor số nguyên cũng phải chạy được"
print("✅ Bài 12 đúng")
```
